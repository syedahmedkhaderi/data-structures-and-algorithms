```python
def coinChange(totalNumber, coins):
    coins.sort
    index = len(coins)-1
    N = totalNumber

    while True:
        curr_coin = coins[index]

        if N >= curr_coin:
            print(curr_coin)
            N = N - curr_coin
         
        if N < curr_coin:
            index -= 1
        
        if N == 0:
            break
    
coins = [1,2,5,20,50,100]
coinChange(201, coins)
```
