```python
class Item:
    def __init__(self, weight, value):
        self.weight = weight
        self.value = value
        self.ratio = value / weight


def knapsackMethod(items, capacity):
    used_capacity = 0
    total_val = 0
    items.sort(key = lambda x:x.ratio, reverse = True) # descending order of ratio to get smallest value item first

    for item in items:
        if used_capacity + item.weight <= capacity:
            used_capacity += item.weight
            total_val += item.value
        else:
            unused_capacity = capacity - used_capacity
            value = item.ratio * unused_capacity
            
            used_capacity += unused_capacity
            total_val += value
        if used_capacity == capacity:
            break

    print("Total value obtained: "+str(total_val))
```

```python
item1 = Item(20,100)
item2 = Item(30,120)
item3 = Item(10,60)
cList = [item1, item2, item3]

knapsackMethod(cList, 50)
```
