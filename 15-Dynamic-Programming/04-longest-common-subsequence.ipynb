### Longest Common Subsequence, Top Down DP (Memoization)
- Given two strings, find the length of their longest common subsequence (characters in the same relative order, not necessarily contiguous).
- Walk both strings from the start: if the current characters match, they must be part of the LCS, so add 1 and advance both indices. Otherwise, try advancing each string by one and take the better result.
- Uses Top-Down Dynamic Programming (Memoization) by storing subproblem results indexed by `(index1, index2)` in `memo`.

How it's implemented:
- Base case: either index reaches the end of its string -> no more characters to match, return 0.
- Check if `(index1, index2)` is in `memo`: if not, calculate:
  - If `s1[index1] == s2[index2]`: `1 + findLCS(index1+1, index2+1, memo)`.
  - Otherwise: `max(findLCS(index1, index2+1, memo), findLCS(index1+1, index2, memo))`.
- Return `memo[(index1, index2)]`.

```python
def findLCS(s1, s2, index1, index2, memo):
    if index1 == len(s1) or index2 == len(s2):
        return 0
    dictKey = (index1, index2)
    if dictKey not in memo:
        if s1[index1] == s2[index2]:
            memo[dictKey] = 1 + findLCS(s1, s2, index1 + 1, index2 + 1, memo)
        else:
            op1 = findLCS(s1, s2, index1, index2 + 1, memo)
            op2 = findLCS(s1, s2, index1 + 1, index2, memo)
            memo[dictKey] = max(op1, op2)
    return memo[dictKey]
```

```python
print(findLCS("elephant", "eretpat", 0, 0, {}))
```
