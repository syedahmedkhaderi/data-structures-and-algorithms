### Minimum Cost Path in a 2D Array, Top Down DP (Memoization)
- Given a 2D grid of costs, find the minimum total cost to travel from the top-left cell `(0,0)` to a target cell `(row, col)`.
- Movement is restricted to right and down, so recursively the cheapest way to reach `(row, col)` is the cheaper of arriving from above or from the left, plus the cost of the current cell.
- Uses Top-Down Dynamic Programming with Memoization (`memo`) indexed by `(row, col)`.

How it's implemented:
- Base case: if `row` or `col` is `-1` (off the grid), that path is invalid -> return `infinity` so it's never chosen by `min`.
- Base case: `(0,0)` costs just `twoDArray[0][0]`.
- Check if state `(row, col)` is in `memo`: if not, compute `twoDArray[row][col] + min(cost from above, cost from left)` and save it.
- Return `memo[(row, col)]`.

```python
def findMinCost(twoDArray, row, col, memo):
    if row == -1 or col == -1:
        return float('inf')
    elif row == 0 and col == 0:
        return twoDArray[0][0]
    else:
        dictKey = (row, col)
        if dictKey not in memo:
            op1 = findMinCost(twoDArray, row - 1, col, memo)
            op2 = findMinCost(twoDArray, row, col - 1, memo)
            memo[dictKey] = twoDArray[row][col] + min(op1, op2)
        return memo[dictKey]
```

```python
TwoDList = [[4,7,8,6,4],
           [6,7,3,9,2],
           [3,8,1,2,4],
           [7,1,7,3,7],
           [2,9,8,9,3]
           ]

print(findMinCost(TwoDList, 4, 4, {}))
```
