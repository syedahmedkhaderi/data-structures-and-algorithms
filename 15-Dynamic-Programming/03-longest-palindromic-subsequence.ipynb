### Longest Palindromic Subsequence, Top Down DP (Memoization)
- Find the length of the longest subsequence of a string that is a palindrome (characters don't need to be contiguous).
- Compare the characters at the two ends of the current window: if they match, they can be the outer pair of a palindrome, so add 2 and recurse inward. If not, try shrinking the window from either side and take the best result.
- Uses Top-Down Dynamic Programming with Memoization (`memo`) indexed by `(startIndex, endIndex)`.

How it's implemented:
- Base case: `startIndex > endIndex` -> empty window, length 0.
- Base case: `startIndex == endIndex` -> single character, length 1.
- Check if state `(startIndex, endIndex)` is in `memo`: if not, compute it:
  - If outer characters match: `2 + findLPS(start+1, end-1, memo)`.
  - Otherwise: `max(findLPS(start, end-1, memo), findLPS(start+1, end, memo))`.
- Return `memo[(startIndex, endIndex)]`.

```python
def findLPS(s, startIndex, endIndex, memo):
    if startIndex > endIndex:
        return 0
    elif startIndex == endIndex:
        return 1
    dictKey = (startIndex, endIndex)
    if dictKey not in memo:
        if s[startIndex] == s[endIndex]:
            memo[dictKey] = 2 + findLPS(s, startIndex + 1, endIndex - 1, memo)
        else:
            op1 = findLPS(s, startIndex, endIndex - 1, memo)
            op2 = findLPS(s, startIndex + 1, endIndex, memo)
            memo[dictKey] = max(op1, op2)
    return memo[dictKey]
```

```python
print(findLPS("ELRMENMET", 0, 8, {}))
```
