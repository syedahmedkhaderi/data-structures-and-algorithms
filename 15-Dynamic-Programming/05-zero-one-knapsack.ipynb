### 0/1 Knapsack Problem, Top Down DP (Memoization)
- Given a set of items with a profit and a weight, and a knapsack with a fixed weight `capacity`, find the maximum profit achievable without exceeding the capacity.
- Each item can either be taken (once) or skipped -- that's the "0/1" part. For each item we branch into both choices and keep the better outcome.
- Uses Top-Down Dynamic Programming with Memoization (`memo`) indexed by `(currentIndex, capacity)`.

How it's implemented:
- Base case: no capacity left, or we've run out of items to consider -> profit 0.
- Check if state `(currentIndex, capacity)` is in `memo`: if so, return `memo[(currentIndex, capacity)]`.
- If the current item fits in the remaining capacity: compare taking it (`profit + recurse with reduced capacity`) versus skipping it, store the max in `memo[(currentIndex, capacity)]` and return.
- If the current item doesn't fit: forced to skip, store 0 and return.

```python
class Item:
    def __init__(self, profit, weight):
        self.profit = profit
        self.weight = weight

def zoKnapsack(items, capacity, currentIndex, memo):
    dictKey = (currentIndex, capacity)
    if capacity <= 0 or currentIndex < 0 or currentIndex >= len(items):
        return 0
    elif dictKey in memo:
        return memo[dictKey]
    elif items[currentIndex].weight <= capacity:
        profit1 = items[currentIndex].profit + zoKnapsack(items, capacity - items[currentIndex].weight, currentIndex + 1, memo)
        profit2 = zoKnapsack(items, capacity, currentIndex + 1, memo)
        memo[dictKey] = max(profit1, profit2)
        return memo[dictKey]
    else:
        memo[dictKey] = 0
        return 0
```

```python
mango = Item(31, 3)
apple = Item(26, 1)
orange = Item(17, 2)
banana = Item(72, 5)

items = [mango, apple, orange, banana]

print(zoKnapsack(items, 7, 0, {}))
```
