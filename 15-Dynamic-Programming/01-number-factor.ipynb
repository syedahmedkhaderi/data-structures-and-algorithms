### Number Factor Problem, Top Down DP (Memoization)
- Count the number of ways to write a number `n` as an ordered sum of 1s, 3s, and 4s using Top-Down Dynamic Programming with Memoization.
- The last term added to reach `n` was either a 1, a 3, or a 4, so the total ways to reach `n` is the sum of the ways to reach `n-1`, `n-3`, and `n-4`.
- Store subproblem results in a dictionary (`memo`) to avoid recomputing already solved subproblems.

How it's implemented:
- Base cases: `n` in `{0, 1, 2}` -> 1 way; `n == 3` -> 2 ways (these seed the recursion).
- Check if state `n` is already stored in `memo`: if so, return `memo[n]`.
- General case: compute `numberFactor(n-1, memo) + numberFactor(n-3, memo) + numberFactor(n-4, memo)`, store in `memo[n]`, and return.

```python
def numberFactor(n, memo):
    if n in (0,1,2):
        return 1
    elif n == 3:
        return 2
    else:
        if n not in memo:
            subP1 = numberFactor(n-1, memo)
            subP2 = numberFactor(n-3, memo)
            subP3 = numberFactor(n-4, memo)
            memo[n] = subP1 + subP2 + subP3
        return memo[n]
```

```python
print(numberFactor(5, {}))
```
