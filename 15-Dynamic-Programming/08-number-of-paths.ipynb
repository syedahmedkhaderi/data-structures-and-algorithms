### Number of Paths in a 2D Array (with target cost), Top Down DP (Memoization)
- Given a 2D grid of costs, count how many distinct paths from the top-left cell to a given `(row, col)` cell add up to exactly a given `cost`.
- Movement is restricted to right and down (equivalently, recursion moves from `(row, col)` back toward `(0,0)` by going up or left).
- At every cell we subtract that cell's value from the remaining `cost` budget, and recurse into the cell above and the cell to the left, summing the number of valid paths from each.
- Uses Top-Down Dynamic Programming with Memoization (`memo`) indexed by state `(row, col, cost)`.

How it's implemented:
- Base case: if `cost` goes negative, that path is invalid -> 0 paths.
- Base case: if we're at `(0,0)`, it's a valid path only if the remaining cost exactly matches `twoDArray[0][0]`.
- Check if state `(row, col, cost)` is stored in `memo`: if not, compute:
  - Edge cases: if `row == 0` we can only come from the left; if `col == 0` we can only come from above.
  - General case: sum the paths coming from above (`row-1, col`) and from the left (`row, col-1`).
- Return `memo[(row, col, cost)]`.

```python
def numberOfPaths(twoDArray, row, col, cost, memo):
    if cost < 0:
        return 0
    elif row == 0 and col == 0:
        if twoDArray[0][0] - cost == 0:
            return 1
        else:
            return 0
    dictKey = (row, col, cost)
    if dictKey not in memo:
        if row == 0:
            memo[dictKey] = numberOfPaths(twoDArray, 0, col - 1, cost - twoDArray[row][col], memo)
        elif col == 0:
            memo[dictKey] = numberOfPaths(twoDArray, row - 1, 0, cost - twoDArray[row][col], memo)
        else:
            op1 = numberOfPaths(twoDArray, row - 1, col, cost - twoDArray[row][col], memo)
            op2 = numberOfPaths(twoDArray, row, col - 1, cost - twoDArray[row][col], memo)
            memo[dictKey] = op1 + op2
    return memo[dictKey]
```

```python
TwoDList = [[4,7,1,6],
           [5,7,3,9],
           [3,2,1,2],
           [7,1,6,3]
           ]

print(numberOfPaths(TwoDList, 3, 3, 25, {}))
```
