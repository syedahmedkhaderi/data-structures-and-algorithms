## Disjoint Set (Union-Find)

A disjoint set tracks a collection of items split into non-overlapping groups,
and supports two operations efficiently:

- `find(x)` — which group does `x` belong to (returns the group's root)
- `union(x, y)` — merge the groups containing `x` and `y`

Internally it is just one dictionary, `parent`, where every item points one
step toward its group's root. A `rank` dictionary tracks an upper bound on
tree height, used to keep trees shallow when merging.

```python
class DisjointSet:

    def __init__(self, vertices):
        self.vertices = vertices
        self.parent = {i:i for i in self.vertices}
        self.rank = dict.fromkeys(vertices, 0)

    def find(self, val):
        if self.parent[val] == val:
            return val
        else: # Suppose parent = {"A":"A", "B":"A", "C":"B", "D":"C"} and you call find("D"). So, D belongs to A as we recurse.
            return self.find(self.parent[val])

    def union(self, x, y):
        xroot = self.find(x)
        yroot = self.find(y)

        if self.rank[xroot] < self.rank[yroot]:
            self.parent[xroot] = yroot
        elif self.rank[xroot] > self.rank[yroot]:
            self.parent[yroot] = xroot
        else:
            self.parent[yroot] = xroot
            self.rank[xroot] += 1
```

```python
ds = DisjointSet(["A", "B", "C", "D", "E"])
print("initial parent:", ds.parent)

ds.union("A", "B")
ds.union("A", "C")

print("find(C):", ds.find("C"))
print("B and C connected:", ds.find("B") == ds.find("C"))
print("C and D connected:", ds.find("C") == ds.find("D"))
print("final parent:", ds.parent)
print("final rank:  ", ds.rank)
```
