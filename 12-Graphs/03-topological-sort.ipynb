### Topological Sort, graph algorithm
- used only on a Directed Acyclic Graph (DAG). 
- Just remember the slack, es, ef, ls, lf from software project management class. Basically imagine to traverse the nodes while respecting the dependencies of the other nodes. you cant go to a node unless you have traversed all the nodes before it or nodes its dependent on.

Ex: imagine a graph like this, a->[c] b->[d,e] c->[b]

then the topological sort would be **a c b d e**

How its implemented:
- DFS + Stack
- Run DFS.
- After visiting all neighbors, push the node onto a stack.
- Reverse the stack (or pop from it) to get the topological order.

```python
from collections import defaultdict

class Graph:
    def __init__(self, numberofVertices):
        self.graph = defaultdict(list)
        self.numberofVertices = numberofVertices
    
    def addEdge(self, vertex, edge):
        self.graph[vertex].append(edge)

    '''
    The key idea is:
    1. Visit the current node.
    2. Recursively visit all of its neighbors.
    3. Only after all neighbors are done, put the current node into the result.

    This is called postorder DFS.

    Why stack.insert(0, v)?        (You can also just append and reverse at the end)
    Suppose you have A -> C -> E
    DFS visits: A -> C -> E
    When E finishes: stack = [E]
    Then C finishes: stack = [C, E]
    Then A finishes: stack = [A, C, E]
    Every node is placed before the nodes that depend on it. 

    The DFS helper(util) explores everything reachable from A and fills the stack.
    '''
    def topological_sort_util(self, val, visited, stack):
        visited.append(val)

        for i in self.graph[val]:
            if i not in visited:
                self.topological_sort_util(i, visited, stack)
        stack.insert(0, val)

    def topological_sort(self):
        visited = []
        stack = []

        for node in list(self.graph):
            if node not in visited:
                self.topological_sort_util(node, visited, stack)
        print(stack)
```

```python
customGraph = Graph(8)
customGraph.addEdge("A", "C")
customGraph.addEdge("C", "E")
customGraph.addEdge("E", "H")
customGraph.addEdge("E", "F")
customGraph.addEdge("F", "G")
customGraph.addEdge("B", "D")
customGraph.addEdge("B", "C")
customGraph.addEdge("D", "F")

customGraph.topological_sort()
```
