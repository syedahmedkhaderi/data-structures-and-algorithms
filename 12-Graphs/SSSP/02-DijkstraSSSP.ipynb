### Dijkstra's Algorithm, a graph algorithm 
- used to find the shortest path from one starting node to all other nodes in a weighted graph.


```python
from collections import defaultdict

class Graph:
    def __init__(self):
        self.nodes = set()
        self.edges = defaultdict(list)
        self.distances = {} # Because this is a weighted graph. Storing format - ("A","B") : 2


    '''
    defaultdict(list) automatically creates an empty list if a key does not exist.
    Example:
        edges["A"].append("B")
    becomes:
        edges = {
            "A": ["B"]
        }
    '''

    def addNode(self, value):
        self.nodes.add(value)

    def addEdge(self, node1, node2, distance):
        self.edges[node1].append(node2)
        self.distances[(node1, node2)] = distance
```

```python
def dijkstra(graph, initial):
    visited = {initial: 0}
    path = defaultdict(list)

    nodes = set(graph.nodes)

    while nodes:
        minNode = None
        for node in nodes:
            if node in visited:
                if minNode is None:
                    minNode = node
                elif visited[node] < visited[minNode]:
                    minNode = node
            
        if minNode is None:
            return
        
        nodes.remove(minNode)
        current_weight = visited[minNode]

        for edge in graph.edges[minNode]:
            weight = current_weight + graph.distances[(minNode, edge)]
            if edge not in visited or weight < visited[edge]:
                visited[edge] = weight
                path[edge] = minNode
    
    return visited, path
```

```python
customGraph = Graph()
customGraph.addNode("A")
customGraph.addNode("B")
customGraph.addNode("C")
customGraph.addNode("D")
customGraph.addNode("E")
customGraph.addNode("F")
customGraph.addNode("G")
customGraph.addEdge("A", "B", 2)
customGraph.addEdge("A", "C", 5)
customGraph.addEdge("B", "C", 6)
customGraph.addEdge("B", "D", 1)
customGraph.addEdge("B", "E", 3)
customGraph.addEdge("C", "F", 8)
customGraph.addEdge("D", "E", 4)
customGraph.addEdge("E", "G", 9)
customGraph.addEdge("F", "G", 7)

print(dijkstra(customGraph, "A"))
```
