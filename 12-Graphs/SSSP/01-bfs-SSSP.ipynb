### BFS for finding the shortest path between two nodes in an unweighted graph

```python
class Graph:
    def __init__(self, gdict=None):
        if gdict is None:
            gdict = {}
        self.gdict = gdict
    
    def bfs(self, start, end):
        queue = []
        queue.append([start])
        while queue:
            path = queue.pop(0)
            node = path[-1]

            if node == end:
                return path

            for other_node in self.gdict[node]:
                new_path = list(path) # Make a separate copy of the current path, add this neighbor to that copy, and put it into the queue.
                new_path.append(other_node)
                queue.append(new_path)

customDict = { "a" : ["b", "c"],
               "b" : ["d", "g"],
               "c" : ["d", "e"],
               "d" : ["f"],
               "e" : ["f"],
               "g" : ["f"]
            }

g = Graph(customDict)
print(g.bfs("a", "e"))
```
