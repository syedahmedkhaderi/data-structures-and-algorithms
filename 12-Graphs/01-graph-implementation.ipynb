### Undirected Graph using Adjacency list

```python
class Graph:
    def __init__(self):
        self.adjacency_list = {}

    def insert_node(self, val):
        if val not in self.adjacency_list:
            self.adjacency_list[val] = []
            return f"Successfully Inserted node: {val}"
    
    def insert_edge(self, val1, val2):
        if val1 not in self.adjacency_list:
            print(f"{val1} node does not exist, please insert first \n") 
            return

        if val2 not in self.adjacency_list[val1]:
            self.adjacency_list[val1].append(val2)
            self.adjacency_list[val2].append(val1)
            return f"successfully linked {val1} and {val2}"
        return "edge already exists"
    
    def print_graph(self):
        for val in self.adjacency_list:
            print(val, "-->", " ".join(map(str, self.adjacency_list[val])))
            # print(val, ":", self.adjacency_list[val])

    def remove_edge(self, val1, val2):
        if val1 and val2 in self.adjacency_list:
            try:
                self.adjacency_list[val1].remove(val2)
                self.adjacency_list[val2].remove(val1)
            except ValueError:
                pass
            return
        return "edge not found"
    
    def remove_node(self, val):
        if val in self.adjacency_list:
            for other_nodes in self.adjacency_list[val]:
                self.adjacency_list[other_nodes].remove(val)
            del self.adjacency_list[val]
            return f"Successfully removed {val}"
        return "No such node exists"
```

```python
g = Graph()

# Inserting Nodes first
g.insert_node("A")
g.insert_node("B")
g.insert_node("c")
g.insert_node("d")
g.insert_node("e")

# Adding Edge between them
g.insert_edge("A", "B")
g.insert_edge("A", "c")
g.insert_edge("d", "e")
g.insert_edge("C", "e")
g.insert_edge("B", "c")
g.insert_edge("e", "A")

# Printing the graph
g.print_graph()

# removing vertex/edge
g.remove_edge("B", "c")

# removing node
g.remove_node("d")

print()
g.print_graph()
```
