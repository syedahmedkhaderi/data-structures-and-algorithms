```python
# Placeholder for infinity
INF = 9999

# Printing the solution
def printSolution(nV, distance):
    for i in range(nV):
        for j in range(nV):
            if(distance[i][j] == INF):
                print("INF", end=" ")
            else:
                print(distance[i][j], end="  ")
        print(" ")


# Row = FROM and Column = TO. so,G[0][1] means A → B
# nV = number of vertices ; G = graph matrix
def floydWarshall(nV, G):
    distance = G
    for k in range(nV): # k means "Which node am I allowed to use as a middle stop?"
        for i in range(nV): # i means starting node
            for j in range(nV): # j means destination node
                distance[i][j] = min(distance[i][j], distance[i][k]+distance[k][j])
    
    printSolution(nV, distance)
```

```python
''' This is how the matric looks like initially

      A   B    C    D
A     0   8   INF   1
B    INF  0    1   INF
C     4  INF   0   INF
D    INF  2    9    1

'''

G = [[0, 8, INF,1],    
     [INF, 0, 1,INF],
     [4, INF, 0,INF],
     [INF, 2, 9,1]
]

floydWarshall(4, G)
```
