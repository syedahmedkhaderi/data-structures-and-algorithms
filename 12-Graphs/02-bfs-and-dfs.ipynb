### Breadth First Search and Depth First Search Implmentation

- bfs - uses queue because it always explores the earliest discovered node first, allowing it to visit nodes level by level (FIFO)
- dfs - uses stack because it always explores the most recently discovered node first, allowing it to go as deep as possible before backtracking (LIFO)

```python
class Graph:
    def __init__(self, gdict=None):
        if gdict is None:
            gdict = {}
        self.gdict = gdict

    def add_edge(self, v1, v2):
        self.gdict[v1].append(v2)


    ''' 
    so basically, you create 2 lists
    1. visited - to keep track of visited nodes to not add them to queue again
    2. queue - uses fifo. we will pop the first element and add its edges to the queue if not visited
    lets begin, 
    first we will loop until we have visited all the nodes once
    next, we will pop the first element of queue and print it
    then, we will loop through its vertex 
    if the node has been visited then it simply gets ignored and nothing happens
    but if it isnt then it gets added to the visited and queue

    we do this until we have traveresed through all the nodes once. 
    
    duplicate nodes never get added to the queue since they are ignored due to being visited.
    '''

    def bfs(self, val):
        visited = [val]
        queue = [val]

        while queue:
            node = queue.pop(0)
            print(node)
            for other_vertex in self.gdict[node]:
                if other_vertex not in visited:
                    visited.append(other_vertex)
                    queue.append(other_vertex)

    # Same code applies for dfs but we use stack instead of 
    def dfs(self, val):
        visited = [val]
        stack = [val]

        while stack:
            node = stack.pop()
            print(node)
            for other_vertex in self.gdict[node]:
                if other_vertex not in visited:
                    visited.append(other_vertex)
                    stack.append(other_vertex)
```

```python
customDict = { "a" : ["b","c"],
               "b" : ["a", "d", "e"],
               "c" : ["a", "e"],
               "d" : ["b", "e", "f"],
               "e" : ["d", "f", "c"],
               "f" : ["d", "e"]
            }

g = Graph(customDict)
g.dfs("a")
```
