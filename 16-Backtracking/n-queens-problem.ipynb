### N-Queens Problem (Backtracking)
- Place `N` chess queens on an `N x N` chessboard such that no two queens attack each other (no two queens share the same row, column, or diagonal).
- Uses **Backtracking** to incrementally place queens column by column. If placing a queen in a particular row leads to a conflict in subsequent columns, the algorithm backtracks by removing the queen and trying the next row.

How it's implemented:
- `is_place_safe(row_index, col_index)` checks whether placing a queen at `(row_index, col_index)` is safe by checking:
  1. Horizontal row to the left (same row, previous columns).
  2. Upper-left diagonal.
  3. Lower-left diagonal.
- `solve(col_index)` attempts to place a queen in column `col_index` by iterating through rows `0` to `N-1`:
  - If safe, places queen (`chess_table[row_index][col_index] = 1`).
  - Recursively calls `solve(col_index + 1)`.
  - If the recursive call returns `True`, the solution is found.
  - Otherwise, **backtracks** by resetting `chess_table[row_index][col_index] = 0` and trying the next row.
- Base case: when `col_index == N`, all `N` queens are successfully placed.
- **Time Complexity:** $O(N!)$ worst case.
- **Space Complexity:** $O(N^2)$ for the chessboard matrix (plus $O(N)$ recursion stack space).

```python
class NQueens:
    def __init__(self, n):
        self.n = n
        self.chess_table = [[0 for _ in range(n)] for _ in range(n)]

    def print_queens(self):
        for i in range(self.n):
            for j in range(self.n):
                if self.chess_table[i][j] == 1:
                    print(" Q ", end="")
                else:
                    print(" - ", end="")
            print()

    def is_place_safe(self, row_index, col_index):
        # Horizontal check (same row, previous columns)
        for i in range(col_index):
            if self.chess_table[row_index][i] == 1:
                return False

        # Upper-left diagonal check
        i, j = row_index, col_index
        while i >= 0 and j >= 0:
            if self.chess_table[i][j] == 1:
                return False
            i -= 1
            j -= 1

        # Lower-left diagonal check
        i, j = row_index, col_index
        while i < self.n and j >= 0:
            if self.chess_table[i][j] == 1:
                return False
            i += 1
            j -= 1

        return True

    def solve(self, col_index):
        if col_index == self.n:
            return True

        for row_index in range(self.n):
            if self.is_place_safe(row_index, col_index):
                self.chess_table[row_index][col_index] = 1
                if self.solve(col_index + 1):
                    return True
                self.chess_table[row_index][col_index] = 0

        return False

    def solve_NQueens(self):
        if self.solve(0):
            self.print_queens()
        else:
            print("There is no solution for the problem!")
```

```python
queens = NQueens(4)
queens.solve_NQueens()
```
