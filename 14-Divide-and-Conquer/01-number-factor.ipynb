### Number Factor Problem, recursion
- Count the number of ways to write a number `n` as an ordered sum of 1s, 3s, and 4s (a tiling-style counting problem).
- The last term added to reach `n` was either a 1, a 3, or a 4, so the total ways to reach `n` is the sum of the ways to reach `n-1`, `n-3`, and `n-4`.

How it's implemented:
- Base cases: `n` in `{0, 1, 2}` -> 1 way; `n == 3` -> 2 ways (these seed the recursion).
- General case: `numberFactor(n-1) + numberFactor(n-3) + numberFactor(n-4)`.

```python
def numberFactor(n):
    if n in (0,1,2):
        return 1
    elif n == 3:
        return 2
    else:
        subP1 = numberFactor(n-1)
        subP2 = numberFactor(n-3)
        subP3 = numberFactor(n-4)
        return subP1+subP2+subP3
```

```python
print(numberFactor(5))
```
