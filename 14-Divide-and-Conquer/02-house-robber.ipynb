### House Robber Problem, recursion
- Given the amount of money stashed in each house along a street, find the maximum amount that can be robbed without ever robbing two adjacent houses.
- At each house we have two choices: rob it (and skip the next one) or skip it (and consider the next one). We take whichever choice yields more money.

How it's implemented:
- Base case: once `currentIndex` runs past the last house, there's nothing left to rob -> 0.
- `stealFirstHouse`: take this house's money, then jump two houses ahead (skipping the adjacent one).
- `skipFirstHouse`: leave this house, move to the very next one.
- Return `max(stealFirstHouse, skipFirstHouse)`.

```python
def houseRobber(houses, currentIndex):
    if currentIndex >= len(houses):
        return 0
    else:
        stealFirstHouse = houses[currentIndex] + houseRobber(houses, currentIndex + 2)
        skipFirstHouse = houseRobber(houses, currentIndex+1)
        return max(stealFirstHouse, skipFirstHouse)
```

```python
houses = [6,7,1,30,8,2,4]
print(houseRobber(houses, 0))
```
